# 07 · Key Findings

The 7 most surprising things we found in 20 years of Hacker News data.

```python
from IPython.display import Image, display as ipy_display
from pathlib import Path

def show(path: str, caption: str = '') -> None:
    p = Path(f'../data/{path}')
    if p.exists():
        ipy_display(Image(str(p), width=900))
    if caption:
        print(f'↑ {caption}\n')
```

## Finding 1 — Technology life cycles are shockingly short

The hype heatmap shows that most technologies go from birth to peak to irrelevance in **3–5 years** on HN. Blockchain peaked in 2018 and is now at background noise. Heroku flatlines after 2022. ChatGPT's spike in 2023 is the single largest one-year jump in the dataset.

```python
show('fig_hype_heatmap.png', 'Technology hype cycles 2007–2024')
```

## Finding 2 — ChatGPT was a 10× discontinuity

Every previous AI milestone (AlphaGo, GPT-3, DALL-E) caused a bump. ChatGPT caused a step-change. In November 2022, AI-related stories went from ~3% of front-page content to ~18% almost overnight — and never fully came back down.

```python
show('fig_ai_discourse.png', 'AI discourse monthly, 2012–2024')
```

## Finding 3 — The best time to post is Tuesday 10am UTC

Weekend and late-night posts underperform by 30–40% in median score. The scoring sweet spot is weekday mornings in US Eastern time (UTC-5), when the most active voters are just starting their day.

```python
show('fig_timing_heatmap.png', 'Median score by day × hour')
```

## Finding 4 — HN score inequality rivals wealth inequality

The Gini coefficient of story scores is above 0.8 — comparable to national wealth inequality in the most unequal countries. The top 1% of stories capture more than 40% of all upvotes ever cast.

```python
show('fig_lorenz.png', 'Lorenz curve of HN story scores')
```

## Finding 5 — GitHub replaced personal blogs as the default link target

In 2010, the top domains were personal `.com` and `.org` blogs. By 2018, `github.com` was the single most linked domain. Show HN posts almost exclusively link to GitHub — it became the de-facto portfolio site for the HN audience.

```python
show('fig_domain_share.png', 'Domain share over time')
```

## Finding 6 — AI stories score 2× higher than average since 2023

Before 2020 there was no significant score premium for AI-related stories. By 2023–2024 the median AI story scores roughly double the non-AI baseline — meaning the topic itself is now a significant upvote signal, independent of quality.

```python
show('fig_ai_score_premium.png', 'Score premium of AI stories')
```

## Finding 7 — A 50-user network drives a disproportionate share of viral content

The top 50 submitters by high-score post count have each posted 10–50+ stories scoring above 200. HN's front page is significantly shaped by a small, recurring set of curators — not random submissions.

```python
show('fig_power_authors.png', 'Top submitters of viral stories')
```
